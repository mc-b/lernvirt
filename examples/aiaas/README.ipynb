AI as a Service
---------------

Die erste virtuelle Maschine `aiaas-lernvirt-vm-0` stellt den KI-Dienst bereit.

Darauf läuft **Ollama als System-Service** und verwaltet **zwei KI-Modelle**, die bedarfsgesteuert geladen und wieder entladen werden. 

Die Modelle werden zentral betrieben und stehen allen Clients ueber eine **OpenAI-kompatible API** zur Verfuegung.

---

Dieser Code zeigt, wie aus einem Python-Jupyter-Notebook ueber die OpenAI-kompatible API auf einen lokal betriebenen Ollama-Service zugegriffen wird.
Die Verbindung erfolgt ueber den lokalen API-Endpoint, wobei der API-Key lediglich als Platzhalter dient.
Die Funktion ask kapselt einen einfachen Chat-Request und ermoeglicht es, unterschiedliche KI-Modelle gezielt anzusprechen, ohne den restlichen Code anpassen zu muessen.

```python
from openai import OpenAI

client = OpenAI(
    base_url="http://localhost:11434/v1",
    api_key="ollama"  # Dummy, wird ignoriert
)

def ask(model, prompt):
    response = client.chat.completions.create(
        model=model,
        messages=[
            {"role": "user", "content": prompt}
        ],
        temperature=0.2,
    )
    return response.choices[0].message.content
```

Dieses Codebeispiel sendet eine Anfrage an das Allround-Sprachmodell llama3.1:8b-instruct-q4_K_M.
Es eignet sich fuer erklaerende und allgemein-technische Fragestellungen und demonstriert eine einfache Textabfrage ueber die OpenAI-kompatible Schnittstelle.

```python
print(ask(
    "llama3.1:8b-instruct-q4_K_M",
    "Erkläre HTTP Status Codes kurz."
))
```

Dieses Codebeispiel verwendet das Modell qwen2.5:7b-instruct-q4_K_M, das sich besonders fuer strukturierte Antworten und Code-nahe Aufgaben eignet.

```python
print(ask(
    "qwen2.5:7b-instruct-q4_K_M",
    "Schreibe ein kurzes Python Beispiel für einen REST Client."
))
```

Beide Codebeispiele zeigen, wie gezielt unterschiedliche Modelle fuer unterschiedliche Aufgaben angesprochen werden koennen.
