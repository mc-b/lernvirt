## Image Verarbeitung

Quelle: [Analyze images and files](https://developers.openai.com/api/docs/quickstart?input-type=file-url&lang=python&language=python#analyze-images-and-files)


```python
import os
from openai import OpenAI
from IPython.display import Markdown, display
from datetime import datetime
import requests
import base64
```

```python
client = OpenAI(
    base_url="http://10.10.1.10:11434/v1",
    api_key="ollama"  # Dummy, wird ignoriert
)

# 1. Bild herunterladen
image_url = "https://api.nga.gov/iiif/a2e6da57-3cd1-4235-b20e-95dcaefed6c8/full/!800,800/0/default.jpg"
response = requests.get(image_url)
response.raise_for_status()

# 2. In Base64 umwandeln
image_bytes = response.content
image_base64 = base64.b64encode(image_bytes).decode("utf-8")

# MIME-Type bestimmen (hier JPEG, da URL .jpg ist)
mime_type = "image/jpeg"
data_url = f"data:{mime_type};base64,{image_base64}"

# 3. An OpenAI übergeben
api_response = client.responses.create(
    model="gemma3:12b",
    input=[
        {
            "role": "user",
            "content": [
                {
                    "type": "input_text",
                    "text": "What teams are playing in this image?"
                },
                {
                    "type": "input_image",
                    "image_url": data_url
                }
            ]
        }
    ]
)

print(api_response.output_text)
```
